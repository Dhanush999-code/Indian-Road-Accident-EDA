```python
df['date'] = pd.to_datetime(df['date'])

df['date'].dtype
```

```python
df['time_hour'] = pd.to_datetime(
    df['time'], format='%H:%M'
).dt.hour

(df['hour'] == df['time_hour']).value_counts()
```

```python
df['actual_day'] = df['date'].dt.day_name()

(df['day_of_week'] == df['actual_day']).value_counts()
```

```python
df['accident_severity'].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='accident_severity')

plt.title('Distribution of Accident Severity')
plt.xlabel('Accident Severity')
plt.ylabel('Number of Accidents')

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='day_of_week')

plt.title('Occurance of accident on days')
plt.xlabel('Days')
plt.ylabel('Number of Accidents')

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(data=df, x='cause',
              order=df['cause'].value_counts().index)

plt.title('Accidents by Cause')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='city',
              order=df['city'].value_counts().index)

plt.title('Number of Accidents by City')
plt.xticks(rotation=30)
plt.show()
```

```python
df['weather'].value_counts().plot(
    kind='bar',
    figsize=(7,4),
    title='Accidents by Weather Condition'
)

plt.ylabel('Number of Accidents')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='hour', color='steelblue')

plt.title('Accidents by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Number of Accidents')

plt.show()
```

```python
df['month'] = df['date'].dt.month

monthly = df.groupby('month').size()

monthly.plot(kind='line', marker='o', figsize=(9,5))

plt.title('Monthly Accident Distribution')
plt.xlabel('Month')
plt.ylabel('Number of Records')
plt.xticks(range(1,13))
plt.grid()
plt.show()
```

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(x='visibility', hue='accident_severity', data=df)
plt.title(f'{plot_no}. Visibility Conditions vs Accident Severity')
show_fig()
plot_no += 1
```

CORRELATION ANALYSIS BY USING CORRELATION


```python
numeric_df = df.select_dtypes(include='number')

corr = numeric_df.corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```
